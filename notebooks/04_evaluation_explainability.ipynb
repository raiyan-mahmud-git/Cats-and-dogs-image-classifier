# Stage 4: Comprehensive Evaluation & Explainability

In this notebook, we perform deep diagnostics on our fine-tuned MobileNetV2 model:
1. **Statistical Rigor:** Confusion matrix, Precision, Recall, Specificity, F1-Score, and ROC-AUC on 1,000 validation images.
2. **Calibration:** Reliability diagrams and Expected Calibration Error (ECE).
3. **Sample Size & Confidence Intervals:** Wilson score 95% confidence intervals comparing $N=50$ vs $N=1,000$.
4. **Grad-CAM Heatmaps:** Visual explanations of what features the model focused on for correct predictions and failure cases.

```python
import sys
from pathlib import Path

PROJECT_ROOT = Path("..").resolve()
if str(PROJECT_ROOT / "src") not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT / "src"))

from catdog.config import Config
from catdog.evaluate_deep import run_comprehensive_evaluation

config = Config()
report = run_comprehensive_evaluation(config)
```

## 1. Diagnostic Plots

View the generated confusion matrix, ROC curve, and calibration reliability diagram:

```python
from PIL import Image
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
axes[0].imshow(Image.open(config.reports_dir / "confusion_matrix.png"))
axes[0].axis("off")
axes[1].imshow(Image.open(config.reports_dir / "roc_curve.png"))
axes[1].axis("off")
axes[2].imshow(Image.open(config.reports_dir / "calibration_curve.png"))
axes[2].axis("off")
plt.tight_layout()
plt.show()
```
